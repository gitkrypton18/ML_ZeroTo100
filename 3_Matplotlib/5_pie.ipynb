## 🟣 Pie Chart

### 🧩 Definition
A **Pie Chart** is a circular chart divided into **slices**, where each slice represents the **proportion or percentage** of a category relative to the whole.  
It is mostly used for **categorical data visualization**.

---

### 🔹 Scenarios

#### 1️⃣ Univariate Categorical Data
- Shows the **proportion of each category** in a single variable.
- **Example:** `Market Share of Companies`, `Favorite Fruits of Students`

#### 2️⃣ Numerical vs Categorical (Bi-variate summary)
- Sum or aggregate of numerical values for each category.  
- Each slice shows the **relative contribution** of a category to the total numerical value.
- **Example:** `Total Sales per Product Category`, `Revenue per Region`

#### 3️⃣ Quick Overview of Composition
- Useful when you want to **highlight percentage contribution** rather than exact counts.

---

### 🧠 Summary

| Type of Data | Purpose | Example |
|:-------------:|:---------|:---------|
| Categorical (univariate) | Show proportion of categories | Favorite Fruits |
| Categorical vs Numerical | Show relative contribution to total | Sales per Product |

---

### 💻 Syntax
```python
plt.pie(values, labels=categories, autopct='%1.1f%%', startangle=90, colors=colors)
plt.title('Pie Chart')
plt.show()


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# simple data
data=[23,45,100,20,49]

plt.pie(data)
plt.show()
````

````python
# labelling
data=[23,45,100,20,49]
subjects=['hindi','english','maths','sst','science']
plt.pie(data, labels=subjects)
plt.show()
````

### Dataset gayle


````python
df=pd.read_csv('gayle-175.csv')
df
````

````python
plt.pie(df['batsman_runs'],labels=df['batsman'])
plt.show()
````

````python
# percentage of the runss
# uuse of 'autopct='%0.1f%%'
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
````

````python
#colors
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','red','cyan','pink','green','yellow'])
plt.show()
````

````python
# explode shadow function

plt.pie(df['batsman_runs'], labels=df['batsman'], autopct='%0.1f%%', explode=[0.3,0,0,0,0,0.1] ,shadow=True)
plt.show()
````
