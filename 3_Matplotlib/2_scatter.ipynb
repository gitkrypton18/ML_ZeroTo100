## 🔵 Scatter Plot

### 🧩 Definition
A **Scatter Plot** is used to visualize the **relationship or correlation** between **two numerical variables**.  
Each point on the graph represents a data observation.

---

### 🔹 Scenarios

#### 1️⃣ Both Columns are Numerical
- Used for **bivariate analysis** to check how one variable affects another.  
- **Example:** `Height vs Weight`

#### 2️⃣ Identify Correlation
- Positive, Negative, or No correlation between variables.  
- **Example:** `Hours_Studied vs Marks`

#### 3️⃣ Detect Outliers or Patterns
- Helps find **clusters**, **trends**, or **anomalies** in the dataset.  
- **Example:** `Temperature vs IceCream_Sales`

---

### 🧠 Summary
| Type of Data | Purpose | Example |
|:-------------:|:---------|:---------|
| Numerical vs Numerical | Analyze correlation or trend | Height vs Weight |
| Numerical vs Numerical | Detect outliers & patterns | Temperature vs Sales |

---

### 💻 Syntax
```python
plt.scatter(x, y, color='r', marker='o')
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('Scatter Plot')
plt.grid(True)
plt.show()


"Note 2D plot is a special case of scatter plot"

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# plt.scatter simple function
x=np.linspace(-10,10,50)

y=10*x + 3 +  np.random.randint(0,300,50)
y
````

````python
plt.scatter(x,y)
# scatter plot
# correlation
````

### USing Dataset
batter.csv


````python
df=pd.read_csv('batter.csv')
df
````

````python
df=df.head(50)
plt.scatter(df['avg'],df['strike_rate'])
````

````python
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
````

````python
# size parameter
# tips is a default dataset oin seaborn library

tips=sns.load_dataset('tips')
tips
````

````python
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
# the size is telling how many customers came with how many people
# gives us a 3d overview of that information

````

### plt.plot() v/s plt.scatter()

````python
# scatterplot using plt.plot
plt.plot(tips['total_bill'],tips['tip'])
# gives us a 2d plot
````

````python
# scatterplot using plt.plot
plt.plot(tips['total_bill'],tips['tip'],'o')
# gives us a scatterrplot
# faster technique for big datasets
````
