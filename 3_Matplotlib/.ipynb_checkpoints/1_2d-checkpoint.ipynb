## 📈 2D Line Plot

### 🧩 Definition
A **2D Line Plot** is used for **bivariate analysis**, i.e., analyzing the **relationship between two variables**.

---

### 🔹 Scenarios

#### 1️⃣ Both Columns are Numerical
- Used to study the **trend** or **correlation** between two numeric variables.  
- **Example:** `Marks vs Hours_Studied`

#### 2️⃣ One Column is Categorical & Other is Numerical
- Used to show **changes** or **comparison** across categories.  
- **Example:** `Month vs Sales`

#### 3️⃣ Time Series Data
- Ideal for **visualizing trends over time**, where the X-axis represents **Date/Time**.  
- **Example:** `Date vs Stock_Price`

---

### 🧠 Summary
| Type of Data | Purpose | Example |
|:-------------:|:---------|:---------|
| Numerical vs Numerical | Shows correlation or trend | Hours vs Marks |
| Categorical vs Numerical | Compares categories | Month vs Sales |
| Time Series | Tracks change over time | Date vs Stock_Price |

```python
# Multiple plotting 
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
# just write it for Rohit Also

```

### Labelling in Graphs


```python
# Labels Title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
```

```python
# colors(hex) in line(width and style) and marker(size)
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red')

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
```

```python

# line style 
# solid - , dashed----, dotted, dashed dotted

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot')

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
```

```python
# line width

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot',linewidth=2)

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
```

```python
# points and markers and manipulating markers

plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot',linewidth=2,marker='o',markersize=10)

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
```

```python
# legend -> location
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='dashdot',linewidth=2,marker='o',markersize=10,label='Rohit')

plt.title('Rohit Sharma V/S Virat Kohli Career Comparision')
plt.xlabel('Seasson')
plt.ylabel('Runs Scored')
plt.legend()
# legend hass default parameter of location loc='best'
#you can change also the location plt.legend(loc='right'), upper right

```

```python
# whenever we use larger values in this scenario all the small variations are trailed
# and if you want to use variation 
# limiting axis 
price=[48000,54000,57000,49000,47000,45000,5000000]
year=[2015,2016,2017,2018,2019,2020,2025]
plt.plot(year,price)
```

```python

# and if you want to use variation 
# limiting axis 
price=[48000,54000,57000,49000,47000,45000,5000000]
year=[2015,2016,2017,2018,2019,2020,2025]

plt.ylim(0,75000)
plt.plot(year,price)
plt.xlim(2017,2019)
# graph is trimmed
# outliers in graph and limiting axes

# grids in matplotlib
plt.grid()
```
