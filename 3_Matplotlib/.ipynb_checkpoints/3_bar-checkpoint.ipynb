## 🟩 Bar Chart

### 🧩 Definition
A **Bar Chart** is used to represent **categorical data** with **rectangular bars**,  
where the **length or height of each bar** corresponds to the **value or frequency** of that category.

---

### 🔹 Scenarios

#### 1️⃣ Compare Categories
- Best for comparing **different groups or categories**.
- **Example:** `Sales by Region`, `Students per Branch`

#### 2️⃣ Visualize Frequencies or Counts
- Shows how often a particular category occurs.  
- **Example:** `Number of Customers per City`

#### 3️⃣ Analyze Categorical vs Numerical Data
- X-axis → **Categorical variable**  
- Y-axis → **Numerical variable (like sum, average, count, etc.)**

---

### 🧠 Summary

| Type of Data | Purpose | Example |
|:-------------:|:---------|:---------|
| Categorical vs Numerical | Compare categories | Sales by Product |
| Categorical vs Numerical | Show counts or averages | Customers per City |

---

### 💻 Syntax
```python
plt.bar(x, height, color='skyblue', edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Chart')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# simple bar chart
children=[10,20,40,10,30]
colors = ['red', 'black', 'blue', 'green', 'yellow']

plt.bar(colors,children)
````

````python
plt.bar(colors,children,color='black')
````

````python
# horixontal bar
plt.barh(colors,children)
````

### Using Dataset batsman season record

````python
# color and label

df=pd.read_csv('batsman_season_record.csv')
df
````

````python
plt.bar(df['batsman'], df['2015'])
````

````python
plt.bar(df['batsman'], df['2015'])
plt.bar(df['batsman'], df['2016'])
plt.bar(df['batsman'], df['2017'])
# very bad behavior 
# what is this
````

````python
# side by side plotting
plt.bar(df['batsman'],df['2015'],width=0.5)
````

````python
# shape
plt.bar(np.arange(df.shape[0]),df['2015'],width=0.75)
````

````python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='blue')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='green')



# xticks

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
````

````python
# what if the categorical name is very bog 
# using the parameters of the xticks rotation

# back to the basic problem
children=[10,20,40,25,30]
colors=['rediiiiiiiiii','greeeeeniiii','bluieeeee','purpleeeeeee','oraangee']
plt.bar(colors,children, color='black')
````

````python
children=[10,20,40,25,30]
colors=['rediiiiiiiiii','greeeeeniiii','bluieeeee','purpleeeeeee','oraangee']
plt.bar(colors,children, color='black')
plt.xticks(rotation='vertical')
````

````python
# Stacked bar chart: 2015 + 2016 + 2017
plt.bar(df['batsman'], df['2015'], label='2015')
plt.bar(df['batsman'], df['2016'], bottom=df['2015'], label='2016')
plt.bar(df['batsman'], df['2017'], bottom=df['2015']+df['2016'], label='2017')

# Optional enhancements
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.title('Runs by Batsman (2015-2017)')
plt.legend()
plt.show()
````
