## 🟠 Histogram

### 🧩 Definition
A **Histogram** is used to visualize the **distribution of a single (univariate) numerical variable**.  
It shows how **frequently each range (bin) of values occurs** in the dataset.

---

### 🔹 Scenarios

#### 1️⃣ Frequency Distribution
- To see how data is **spread across different value ranges**.
- **Example:** `Marks of Students`, `Age of Customers`

#### 2️⃣ Detect Skewness
- Identify if data is **symmetric, left-skewed, or right-skewed**.

#### 3️⃣ Detect Outliers or Gaps
- Helps find **rare occurrences** or **empty value ranges**.

#### 4️⃣ Summarize Large Data
- Useful when working with **large datasets** to quickly visualize patterns.

---

### 🧠 Summary

| Type of Data | Purpose | Example |
|:-------------:|:---------|:---------|
| Numerical (univariate) | Frequency distribution | Student Marks |
| Numerical (univariate) | Identify skewness & spread | Customer Age |
| Numerical (univariate) | Detect outliers | Sales Amounts |

---

### 💻 Syntax
```python
plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()



````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
# simple data
data= [32,45,56,10,15,27,61]
plt.hist(data,bins=[10,25,40,55,70])
````
