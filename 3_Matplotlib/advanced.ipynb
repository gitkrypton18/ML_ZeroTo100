## 🎨 Colored Scatter Plots in Matplotlib

A **colored scatter plot** is an advanced version of a normal scatter plot in which each point’s **color represents an additional variable**.  
It helps us visualize **three dimensions of data** — X, Y, and Color — on a 2D plane.  
This makes it extremely useful for identifying trends, patterns, or groupings within data.

---

### 🧩 Basic Syntax
```python
plt.scatter(x, y, c=color_values, cmap='colormap_name', s=size, alpha=value, marker='o', edgecolor='color')


We are currently using IRIS Dataset

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
iris=pd.read_csv('iris.csv')
iris.head()
````

````python
# Correlation of Sepal Length and Petal Length
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
````

Color Assignment


````python
# I'm replacing the Sepcies name with a number
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
````

````python
# Now we will use the hidden parameter c=colors for plotting the graph with colors

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
````

````python
# cmap parameter to summon colors
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'], cmap='winter')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
````

````python
## color bar function
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'], cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
````

````python
# Graph Size 

plt.figure(figsize=(15,7))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'], cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
````

````python
## one more excaptional hitters 



plt.figure(figsize=(18,7))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

# new syntax 
plt.axhline(130,color='red')
plt.axvline(30,color='blue')
plt.axhline(140,color='green') # exceptional hitters
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
````

## 🧩 Subplots in Matplotlib

A **subplot** allows you to display **multiple plots (figures)** in a single window or figure canvas.  
This is extremely useful when you want to **compare multiple datasets**, or show **different views of related data** side-by-side.

Matplotlib provides several ways to create subplots, but the most common ones are:
1. `plt.subplot()` — for simple, quick layouts.  
2. `plt.subplots()` — for more flexible and advanced control.

---


````python

plt.scatter(df['avg'],df['strike_rate'])
````

````python
fig, ax=plt.subplots()

ax.scatter(df['avg'],df['strike_rate'])
````

````python
# title
fig, ax=plt.subplots(figsize=(15,6))
ax.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
ax.set_title("Something")
ax.set_xlabel('Avg')
ax.set_ylabel('strike_rate')
fig.show() # same as plt.show
````

````python
# Dual Graphs

fig,ax=plt.subplots(nrows=2,ncols=1)  # 2 rows 1 column
````

````python
fig
````

````python
ax
````

````python
fig, ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))
# the sharex parameter is used to share their x axis in the graph 

ax[0].scatter(df['avg'],df['strike_rate'],color='r')
ax[1].scatter(df['avg'],df['runs'],color='g')
ax[0].set_title('Avg vs StrikeRate')
ax[1].set_title('Avg vs Runs')


ax[1].set_xlabel('Avg')
ax[1].set_ylabel('Runs')
````

````python
# Quad Plots
fig, ax=plt.subplots(nrows=2,ncols=2,figsize=(10,10))
# the sharex parameter is used to share their x axis in the graph 

ax[0,0].scatter(df['avg'],df['strike_rate'],color='r')
ax[0,1].scatter(df['avg'],df['runs'],color='g')
ax[1,0].hist(df['avg'],color='b')
ax[1,1].hist(df['runs'],color='m')


````

````python
# another way of plotting
fig=plt.figure()
ax1=fig.add_subplot(2,1,1)
ax1.scatter(df['avg'],df['strike_rate'],color='r')

ax2=fig.add_subplot(2,1,2)
ax2.scatter(df['avg'],df['runs'])
````

````python
fig=plt.figure()
ax1=fig.add_subplot(1,2,1)
ax1.scatter(df['avg'],df['strike_rate'],color='r')

ax2=fig.add_subplot(1,2,2)
ax2.scatter(df['avg'],df['runs'])
````

````python
# 4 plots
fig=plt.figure()
ax1=fig.add_subplot(2,2,1)
ax1.scatter(df['avg'],df['strike_rate'],color='r')

ax2=fig.add_subplot(2,2,2)
ax2.scatter(df['avg'],df['runs'],color='g')

ax3=fig.add_subplot(2,2,3)
ax3.hist(df['avg'],color='k')

ax4=fig.add_subplot(2,2,4)
ax4.hist(df['runs'],color='m')
````

## 3D Surface plots


````python
# 3d graph plotter we want to plot 
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

````

````python
x
````

````python
xx, yy=np.meshgrid(x,y)
xx.shape
````

````python
yy.shape
````

````python
xx, yy
````

````python
z=xx**2 + yy**2
z.shape
````

````python
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')

p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````

````python
# 4=3d sinx + cosx
z=np.sin(xx) + np.cos(yy)
fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````

## Countour Plot : 


````python
fig=plt.figure(figsize=(12,8))
z=xx**2 + yy**2
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````

````python
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````
