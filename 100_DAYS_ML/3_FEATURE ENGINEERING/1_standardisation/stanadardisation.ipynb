# Understanding `train_test_split`

Mastering the `train_test_split` is a foundational step in preparing data for any Machine Learning model. Here is an in-depth breakdown of the concept, the function, and how to apply it.

## Part 1: The Concept of Train-Test Split

In machine learning, the goal is to build a model that generalizes well to **new, unseen data**. If a model is trained and evaluated on the exact same dataset, it might just memorize the answers (a problem known as overfitting). It will look like it's performing perfectly, but it will fail in the real world. 

To prevent this, data is split into two distinct chunks:

1. **Training Set:** The "practice exam." The model uses this data to learn the relationships between the features and the target.
2. **Testing Set:** The "final exam." This data is hidden from the model during training. After training, the model predicts answers for this set to evaluate its real-world performance.



---

## Part 2: The `train_test_split` Function & Parameters

The `train_test_split` function from Scikit-Learn (`sklearn`) is the industry standard for splitting data quickly and randomly. Here are the conventional parameters you will use:

* `*arrays`: The actual datasets to split. Usually, you pass in `X` (your features/inputs) and `y` (your target/output label).
* `test_size`: The proportion of the dataset to include in the test split.
    * **Float** (e.g., `0.3`): Represents a percentage (30% test, 70% train).
    * **Int** (e.g., `150`): Represents the exact number of rows in the test set.
* `train_size`: The counterpart to `test_size`. Usually, you only need to define one or the other.
* `random_state`: Controls the shuffling process. Setting this to a specific integer (like `0` or `42`) ensures the data splits the **exact same way** every time the code is run, which is crucial for reproducibility and debugging.
* `shuffle` (default=`True`): Determines whether to shuffle the data before splitting. This is almost always desired to prevent the model from learning unintended sorting patterns.
* `stratify` (default=`None`): Determines how the data is sampled. (See Part 4 below for an in-depth explanation).

---

```python
from sklearn.model_selection import train_test_split

# importing tts fromn skleaRN MODEL LIBRARY
```

```python
X_train, X_test, Y_train, Y_test=train_test_split(df.drop('Purchased',axis=1),df['Purchased'],test_size=0.3,random_state=0)   

# means
```

```python
X_train.shape, X_test.shape
```

# Standard Scaler


```python
from sklearn.preprocessing import StandardScaler 

scaler=StandardScaler()

#fit the scaler to the ratio set. it will learmn the parameters 
scaler.fit(X_train) # CALCULATES THE MEAN 

# TRANSSFORM train and test sets
X_train_scaled=scaler.transform(X_train)  # scaling train dataset also
X_test_scaled=scaler.transform(X_test)    # scaling test data also

# we scale both the datset but we do tts on the train data only

#problem : if we give dataframe to standard scaler it returns numpy array

```

```python
scaler.mean_
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)

# now see both train and scaled 
# here we converterd into dataframe 
```

## Effect of Scaling

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

ax1.scatter(X_train['Age'], X_train['EstimatedSalary'])
ax1.set_title("Before Scaling")

ax2.scatter(X_train_scaled['Age'], X_train_scaled['EstimatedSalary'], color='red')
ax2.set_title("After Scaling")

plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

sns.kdeplot(X_train['Age'],ax=ax1)
sns.kdeplot(X_train['EstimatedSalary'],ax=ax1)
ax1.set_title("Before Scaling")
sns.kdeplot(X_train_scaled['Age'],ax=ax2)
sns.kdeplot(X_train_scaled['EstimatedSalary'],ax=ax2)
ax1.set_title("After Scaling")


plt.show()
```

## Comparision of Distributions

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Age Distribution Before Scaling')
sns.kdeplot(X_train['Age'], ax=ax1)

# after scaling
ax2.set_title('Age Distribution After Standard Scaling')
sns.kdeplot(X_train_scaled['Age'], ax=ax2)
plt.show()
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
ax1.set_title('Salary Distribution Before Scaling')
sns.kdeplot(X_train['EstimatedSalary'], ax=ax1)

# after scaling
ax2.set_title('Salary Distribution Standard Scaling')
sns.kdeplot(X_train_scaled['EstimatedSalary'], ax=ax2)
plt.show()
```

## Why scaling is important?

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
lr_scaled=LogisticRegression()
```

```python
lr.fit(X_train,Y_train)
lr_scaled.fit(X_train_scaled,Y_train)
```

```python
y_pred = lr.predict(X_test)
y_pred_scaled = lr_scaled.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print("Actual",accuracy_score(Y_test,y_pred))
print("Scaled",accuracy_score(Y_test,y_pred_scaled))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python

dt = DecisionTreeClassifier()
dt_scaled = DecisionTreeClassifier()
```

```python
dt.fit(X_train,Y_train)
dt_scaled.fit(X_train_scaled,Y_train)
```

```python

y_pred = dt.predict(X_test)
y_pred_scaled = dt_scaled.predict(X_test_scaled)
```

```python

print("Actual",accuracy_score(Y_test,y_pred))
print("Scaled",accuracy_score(Y_test,y_pred_scaled))
```
