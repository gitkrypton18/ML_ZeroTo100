### Problem 1 
### Outlook-Sunny Temp=Hot Humidity=High, Wind = Weak 
### PLay or not
### Solution

### P(Yes | Sunny,Hot,High,Weak)
= P(Sunny | Yes) * P(Hot | Yes) * P(High | Yes) * P(Weak | Yes) * P(Yes)

### P(No | Sunny,Hot,High,Weak)
= P(Sunny | No) * P(Hot | No) * P(High | No) * P(Weak | No) * P(No)

### Compare both probabilities

### If:
### P(Yes | X) > P(No | X)
### Then → Play = Yes

### Else:
### Play = No

### Use Maximum A Posteriori (MAP) Rule
### Choose the class with maximum posterior probability
