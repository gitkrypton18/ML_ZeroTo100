# Naive Bayes Classifier: The Beginner's Guide
This notebook demonstrates the three main types of Naive Bayes classifiers using Scikit-Learn. 
We will use very simple datasets to see exactly how the data format dictates which model we choose.

```python
# Import all the necessary tools
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Import the three Naive Bayes models
from sklearn.naive_bayes import GaussianNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.naive_bayes import BernoulliNB
```

## 1. Gaussian Naive Bayes (For Continuous Measurements)
**Rule:** Use this when your features are decimals or continuous numbers (like weight, length, voltage).

**The Data:** We will use the famous Iris dataset. We are predicting the type of flower based on the length and width of its petals (which are continuous decimal measurements).

```python
from sklearn.datasets import load_iris

# 1. Load the continuous measurement data
data = load_iris()
X_measurements = data.data    # Features: Petal length/width (decimals)
y_flower_type = data.target   # Labels: The flower species

# 2. Split the data (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(
    X_measurements, y_flower_type, test_size=0.2, random_state=42
)

# 3. Train and Predict with Gaussian
gaussian_model = GaussianNB()
gaussian_model.fit(X_train, y_train)

# 4. Check our accuracy
predictions = gaussian_model.predict(X_test)
accuracy = accuracy_score(y_test, predictions)
print(f"Gaussian Model Accuracy: {accuracy * 100:.2f}%")
```

## 2. Multinomial Naive Bayes (For Word Counts / Frequencies)
**Rule:** Use this when your features are whole numbers representing "counts" (how many times something appears). 

**The Data:** We will classify project ideas as either **Fintech (1)** or **Standard Software (0)**. We use a `CountVectorizer` to simply count how many times specific words appear in the text.

```python
from sklearn.feature_extraction.text import CountVectorizer

# 1. Simple text dataset
project_descriptions = [
    "banking transaction anomaly detection",   # Fintech
    "financial trading and investment app",    # Fintech
    "frontend web development using react",    # Standard Software
    "backend database for inventory"           # Standard Software
]
# Labels: 1 = Fintech, 0 = Standard Software
project_labels = [1, 1, 0, 0] 

# 2. Convert the text into word counts
vectorizer = CountVectorizer()
X_word_counts = vectorizer.fit_transform(project_descriptions)

# 3. Train the Multinomial model
multi_model = MultinomialNB()
multi_model.fit(X_word_counts, project_labels)

# 4. Predict on a brand new project description
new_project = ["machine learning for banking security"]
new_counts = vectorizer.transform(new_project)

prediction = multi_model.predict(new_counts)[0]

print(f"Text: '{new_project[0]}'")
print("Prediction:", "Fintech (1)" if prediction == 1 else "Standard Software (0)")
```

## 3. Bernoulli Naive Bayes (For Binary Yes/No Data)
**Rule:** Use this when your features are strictly 1s and 0s (True/False).

**The Data:** Let's imagine we are testing a digital VLSI circuit. We have three binary flags: `[Power Good (1/0), Clock Sync (1/0), Temp OK (1/0)]`. We want to predict if the circuit **Passes (1)** or **Fails (0)** inspection.

```python
# 1. Simple binary dataset
# Columns: [Power, Clock, Temp] -> 1 means OK, 0 means Error
X_circuit_tests = np.array([
    [1, 1, 1],  # Perfect circuit
    [1, 1, 0],  # Temp error, but might still pass
    [0, 1, 1],  # Power error -> Usually fails
    [0, 0, 0]   # Complete failure
])

# Labels: 1 = Circuit Passed, 0 = Circuit Failed
y_circuit_result = np.array([1, 1, 0, 0])

# 2. Train the Bernoulli model
bernoulli_model = BernoulliNB()
bernoulli_model.fit(X_circuit_tests, y_circuit_result)

# 3. Predict on a new circuit coming off the line
# This new circuit has good power (1), but bad clock (0) and bad temp (0)
new_circuit = np.array([[1, 0, 0]]) 

prediction = bernoulli_model.predict(new_circuit)[0]

print("Test Flags: [Power: OK, Clock: FAIL, Temp: FAIL]")
print("Prediction:", "Passed (1)" if prediction == 1 else "Failed (0)")
```
