```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
delivery=pd.read_csv('deliveries.csv')
```

```python
delivery
```

```python
runs=delivery.groupby('batsman').sum()['batsman_runs']
```

```python
runs
```

```python
runs=runs.values
```

```python
sns.distplot(runs)
```

```python
# Sampling on this data
# n=50/461 batsman taken per sample
# m=1000, number pf times sampling done 
```

```python
np.random.choice(runs,size=5)
# pass array and size it will draw array
```

```python
m=1000
n=50
new=[]
for i in range(m):
    x=np.random.choice(runs, size=n)
    new.append(x.mean())
```

```python
len(new)
```

```python
new=np.array(new)
```

```python

sns.distplot(new)
```
