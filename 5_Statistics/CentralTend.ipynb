```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
```

```python
data=pd.read_csv('iris.csv')
data.head()
```

```python
data.rename(columns={
    'SepalLengthCm': 'SL',
    'SepalWidthCm': 'SW',
    'PetalLengthCm': 'PL',
    'PetalWidthCm': 'PW'
}, inplace=True)
```

```python
print("Setosa PL Mean:", np.mean(iris_setosa['PL']))
print("Versicolor PL Mean:", np.mean(iris_versicolor['PL']))
print("Virginica PL Mean:", np.mean(iris_virginica['PL']))
```

```python
print("Setosa PW Mean:", np.mean(iris_setosa['PW']))
print("Versicolor PW Mean:", np.mean(iris_versicolor['PW']))
print("Virginica PW Mean:", np.mean(iris_virginica['PW']))
```

## 2. Median

The **median** is the middle value of a dataset when the values are arranged in ascending or descending order.  
If the number of observations is even, the median is the average of the two middle values.

Median is **less affected by outliers** compared to mean.


```python
arr=np.array([1,4,5,8,12,600,700,800,100000])
np.median(arr)
```


## 3. Mode

The **mode** is the value that occurs **most frequently** in a dataset.

A dataset may have:
- **One mode** (unimodal)
- **More than one mode** (multimodal)
- **No mode** (if all values occur equally)

Mode is useful for **categorical data** and can also be applied to numerical data.


```python
from scipy import stats

print(stats.mode(iris_setosa['PL']))
print(stats.mode(iris_versicolor['PL']))
print(stats.mode(iris_virginica['PL']))
```

## Which one should you use
# Measures of Central Tendency

Central tendency describes a **single value** that represents the **center** of a dataset.

---

## 1. Mean
**Definition:**  
Mean is the arithmetic average of all observations.

**Formula:**  
Mean = (Sum of all values) / (Number of values)

**Use when:**  
- Data is symmetric
- No extreme values (outliers)

**Limitation:**  
- Highly affected by outliers

---

## 2. Median
**Definition:**  
Median is the **middle value** when data is arranged in order.

**Case:**
- Odd values → middle number
- Even values → average of two middle numbers

**Use when:**  
- Data is skewed
- Outliers are present

**Advantage:**  
- Robust to outliers

---

## 3. Mode
**Definition:**  
Mode is the value that occurs **most frequently**.

**Types:**  
- Unimodal (one mode)  
- Bimodal (two modes)  
- Multimodal (multiple modes)

**Use when:**  
- Categorical data
- Identifying most common value

**Limitation:**  
- May not exist or may be multiple

---

## Comparison

| Measure | Affected by Outliers | Best for |
|-------|---------------------|---------|
| Mean | Yes | Balanced numeric data |
| Median | No | Skewed data |
| Mode | No | Categorical data |

---

**Exam Tip:**  
> If data has outliers → use **Median**
