# Measures of Spread

Measures of spread describe **how much the data is spread or varied** around a central value.  
They help us understand the **dispersion** in a dataset.

## Important Measures of Spread
1. **Range**
2. **Interquartile Range (IQR)**
3. **Variance**
4. **Standard Deviation**

---



## 1. Range 
The range describes the difference between the largest and smallest points in your data.

```python
r=np.max(iris_setosa['PL'])-np.min(iris_setosa['PL'])
print('The range is :',r)
```

## 2. Percentiles


A **percentile** is a statistical measure that indicates the value **below which a given percentage of observations** in a dataset falls.

### Example
The **20th percentile** is the value below which **20% of the observations** are found.

### Key Points
- Percentiles divide data into **100 equal parts**
- Used to understand the **relative standing** of a value in a dataset
- Commonly used in **exams, rankings, and performance analysis**


```python
mylist=[1,2,3,4,5,6,7,8,9]
print (np.percentile(mylist,10))
print (np.percentile(mylist,90))
```

```python
np.percentile(iris_setosa['PL'],95)
```

## 3. Quartiles

**Quartiles** divide an ordered dataset into **four equal parts**, each containing **25% of the observations**.


```python
print(np.percentile(iris_setosa['PL'], 25))   # Q1
print(np.percentile(iris_setosa['PL'], 50))   # Q2 (Median)
print(np.percentile(iris_setosa['PL'], 75))   # Q3
print(np.percentile(iris_setosa['PL'], 100))  # Maximum
```

## 4. IQR



```python
iqr=np.percentile(iris_setosa['PL'],75)-np.percentile(iris_setosa['PL'],25)
print('iqr is :',iqr)
```
