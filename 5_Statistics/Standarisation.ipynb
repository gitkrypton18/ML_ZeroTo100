# 1. Standardisation


```python
import numpy as np
import pandas  as pd
import seaborn as sns
```

```python
from scipy.stats import norm
import matplotlib.pyplot as plt
# generate 1000 normal random integers with specified mean and std
draw = norm.ppf(np.random.random(100000),loc=1000,scale=200).astype(int)
sns.distplot(draw)
```

```python
z=[]
for i in draw:
    z.append((i-1000)/200)
```

```python
a=np.array(z)
```

```python
print (a.mean())
print(np.round(a.mean()))
```

```python
print(a.std())
print(np.round(a.std()))
```

```python
sns.distplot(a)
```

# 2 Calculating X Score 


## Step 1: Calculating the Z-Score 
## In this step, we determine how many standard deviations a specific score (700) is from the population mean (500).

```python
# Step 1 : Calculate the z-score
# go to the website x tables : 
z=(700-500)/120
print(z)

# This 0.06 is we get from the z tables
```

### Step 2: Decomposing the Z-score for Table Lookup

Since standard Z-tables only provide values up to two decimal places, we must round and decompose our result:

1. **Rounding:** 1.666... becomes 1.67 (or 1.66 depending on your precision preference).

2. **Splitting for the Table:** To find the area under the curve in a Z-table, split the number into two parts:
    * **The Row (Y-axis):** The first two digits (1.6).
    * **The Column (X-axis):** The hundredths place (0.06).

```python
# So after looking up the  ztable we got a= 0.9515
a=0.9515
```


### Step 3: Finding the Probability (Area)

1. Locate **1.6** in the leftmost column of the Z-table.
2. Move across to the column labeled **.06**.
3. The intersection gives the value **0.9515**.



**Summary of Results:**

* **Z-score = 1.66:** This tells us the score is $1.66$ standard deviations above the mean.
* **Probability = 0.9515:** This tells us that approximately $95.15\%$ of the population scored below this value.
