## Scatter Plots
The scatter plot is a mainstay of statistical visualisation. It depicts the joint distribution of two varibales using a cloud of points, where each point represents an observation in a dataset. The depiction allows the user eye to infer a subastantial amount of information about whether there is any meanningful relationshipn between them

Although, scatter plot can be plotted for both numerical as well as categorical data, it's mostly used to depict relationship between numerical data.

There are 2 ways to draw a scatterplot in seaborn

1. **relplot** [Figure Level Function]

2. **scatterplot** [Axes Level Function]

As we will see, these visualizations can be quite helpful because they use simple and easily-understood representations of data even on complex data by mapping up to three additional variables using the semantics of hue, size, and style.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.style.use('fivethirtyeight')
```

```python
data=sns.load_dataset('tips')
```

```python
data
```

```python
data.head()
```

### Plot a scatterplot between total_bill and tip


```python

# Plot here 
sns.relplot(x='total_bill',y='tip',data=data)

# there is a hidden parameter kind by default kind='scatter'

```

### The hue parameter

```python
# Based on the smoker or not
sns.relplot(x='total_bill',y='tip',hue='smoker',data=data)
```

### The style parameter


```python
sns.relplot(x='total_bill',y='tip',hue='smoker',style='sex',data=data)
```

### The size parameter


```python
sns.relplot(x='total_bill',y='tip',hue='smoker',style='sex',size='size',data=data)
```

### Using the scatterplot function


```python
sns.scatterplot(x='total_bill',y='tip',hue='smoker',data=data)
```

### Categorical Scatterplots


```python

# x-axis -> Categroical Data
# y-axis -> Numerical data
```

```python
plt.style.use('fivethirtyeight')
```

```python
tips=sns.load_dataset('tips')
```

```python

tips.head()
```

```python
### STRIP PLOTS
sns.catplot(x='day',y='tip',kind='strip',data=tips)
# jitter hidden parameter noise using 
```

```python
### SWARM PLOTS
sns.catplot(x='day',y='tip',kind='swarm',data=tips)

## Provides us the distribution of the data 
## the difference between swarm and strip plots
```

```python
# using hue parameter 
sns.catplot(x='day',y='tip',kind='swarm',hue='sex',data=tips)
```

```python
## USING SWARMPLOT
sns.swarmplot(x='day',y='tip',hue='sex',data=tips)
```
