## Scatter Plots
The scatter plot is a mainstay of statistical visualisation. It depicts the joint distribution of two varibales using a cloud of points, where each point represents an observation in a dataset. The depiction allows the user eye to infer a subastantial amount of information about whether there is any meanningful relationshipn between them

Although, scatter plot can be plotted for both numerical as well as categorical data, it's mostly used to depict relationship between numerical data.

There are 2 ways to draw a scatterplot in seaborn

1. **relplot** [Figure Level Function]

2. **scatterplot** [Axes Level Function]

As we will see, these visualizations can be quite helpful because they use simple and easily-understood representations of data even on complex data by mapping up to three additional variables using the semantics of hue, size, and style.

```python
import seaborn as sns

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python

plt.style.use('fivethirtyeight')


```

```python
data=sns.load_dataset('tips')
```

```python
data
```

```python
data.head()
```

### Plot a scatterplot between total_bill and tip


```python

# Plot here 
sns.relplot(x='total_bill',y='tip',data=data)

# there is a hidden parameter kind by default kind='scatter'

```

### The hue parameter

```python
# Based on the smoker or not
sns.relplot(x='total_bill',y='tip',hue='smoker',data=data)
```

### The style parameter


```python
sns.relplot(x='total_bill',y='tip',hue='smoker',style='sex',data=data)
```

### The size parameter


```python
sns.relplot(x='total_bill',y='tip',hue='smoker',style='sex',size='size',data=data)
```

### Using the scatterplot function


```python
sns.scatterplot(x='total_bill',y='tip',hue='smoker',data=data)
```

## BOX PLOT
### Distribution of Data based on a five number summary ('minimum', first quartile(Q1),median,third quartile(Q3) and 'maximum' ). It can tell us the outliers and what their values are. it can also tell us if your data is symmetrical, how tightly your data is grouped and if your data is skewed 




```python
# BOX PLOT :
sns.boxplot(tips['tip'])
```

```python
sns.boxplot(tips['total_bill'])
```

```python
# Categorical Data

sns.catplot(x='day',y='total_bill', kind='box',data=tips)
# now i have a perspective of numerical data using the categgory
```

## HEATMAPS

A heatmap ia a graphical representation of the data where the individuals values contained in a matrix are represented as colors by gradient



```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flight = sns.load_dataset('flights')
flight
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ✅ Load correct dataset again
flight = sns.load_dataset('flights')

# ✅ Check columns to confirm
print(flight.columns)

```

```python
# ✅ Pivot table
x = flight.pivot_table(index='year', columns='month', values='passengers', aggfunc='sum')
print(x.head())


```

```python
# ✅ Heatmap
sns.heatmap(x)
plt.title('Flight Passengers (1949–1960)')
plt.show()
```

```python
sns.heatmap(x,cbar=False)
```

```python
sns.heatmap(x,cbar=False,linewidth=0.5,annot=True,fmt='d')
```

```python
sns.heatmap(x,cbar=False,linewidth=0.5,annot=True,fmt='d',cmap='viridis')
```

## ClusterMap
- A ClusterMap is a type of heatmap in Seaborn that performs hierarchical clustering on both rows and columns of a dataset.
It not only shows the data values as colors but also groups similar rows and columns together using dendrograms (tree diagrams).

- sns.clustermap() creates a heatmap with hierarchical clustering applied to the rows and/or columns, allowing you to visualize patterns, correlations, and groupings in the data.

```python
sns.clustermap(x)
```

```python
9sns.clustermap(x,z_score=0,metric='correlation',annot=True,row_cluster=False)
```
