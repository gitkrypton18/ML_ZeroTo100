# 3. Opening a csv file from an URL

```python
import requests
import pandas as pd
from io import StringIO

# URL of the CSV file hosted online
CSV_URL = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"

# Headers are added to mimic a real browser request
HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) "
        "Gecko/20100101 Firefox/66.0"
    )
}

# Step 1: Send HTTP GET request to fetch the CSV file
response = requests.get(CSV_URL, headers=HEADERS)

# Step 2: Convert the response text into a file-like object
csv_buffer = StringIO(response.text)

# Step 3: Read the CSV data into a pandas DataFrame
df = pd.read_csv(csv_buffer)

# Display the DataFrame
df
```

# 4. Sep Parameter

```python
import pandas as pd

# Path to the TSV (Tab-Separated Values) file
FILE_PATH = "movie_titles_metadata.tsv"

# Column names for the dataset
COLUMN_NAMES = [
    "sno",
    "name",
    "release_year",
    "rating",
    "votes",
    "genres"
]

# Read the TSV file into a pandas DataFrame
df = pd.read_csv(
    FILE_PATH,
    sep="\t",              # TSV files use tab as separator
    names=COLUMN_NAMES     # Explicitly assign column names
)

# Display the DataFrame
df
```

# Renaming 

```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

# 14. Convertors

```python
pd.read_csv('IPL Matches 2008-2020.csv')
```

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```
